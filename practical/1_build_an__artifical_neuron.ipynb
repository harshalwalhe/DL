```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Manually created dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1],
    [2, 1],
    [2, 2]
])

# Step 2: Define weights and bias
weights = np.array([0.5, 0.8])
bias = -0.2

# Step 3: Calculate weighted sum
z = np.dot(X, weights) + bias

# Step 4: Sigmoid activation function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Step 5: Calculate neuron output
output = sigmoid(z)

# Display results
print("Input\t\tWeighted Sum\tOutput")
for i in range(len(X)):
    print(X[i], "\t\t", round(z[i], 2), "\t\t", round(output[i], 4))

# Step 6: Plot input-to-output curve
z_values = np.linspace(-10, 10, 100)
y_values = sigmoid(z_values)

plt.plot(z_values, y_values, label="Sigmoid Function")
plt.scatter(z, output, color="red", label="Dataset Outputs")
plt.title("Artificial Neuron - Sigmoid Activation")
plt.xlabel("Weighted Input (Z)")
plt.ylabel("Neuron Output")
plt.grid(True)
plt.legend()
plt.show()
```
